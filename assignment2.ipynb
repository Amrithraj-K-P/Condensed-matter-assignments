## Question-1

```python
import matplotlib.pyplot as plt
import numpy as np

a=5
# m = 1/2
# h/2pi = 1

def plot_energy(l,c='r'):
    x = np.arange(-15, 15, 0.1)
    y = [ ((i + l*2*np.pi/a))**2 for i in x ]
    plt.plot(x, y, label=f'l={l}',color=c)
for l in range(-3, 4):
    plot_energy(l)  
plt.ylim(0,30)
plt.xlim(-6*np.pi/a,6*np.pi/a)
plt.legend()
plt.grid(True)
plt.xlabel('k')
plt.ylabel('E_k')
plt.axvline(np.pi/a, color='black', lw=1, ls='--')
plt.axvline(-np.pi/a, color='black', lw=1, ls='--')
plt.axvline(0, color='black', lw=1)
plt.show()

''' consider the region (0,pi/a). In this region:
Band index  m
----------|---
0           0
1           1
2          -1
3           2
4          -2
..         ...

So Band index(m) =  -2m        if m<=0
                 =   2m - 1    if m>0
'''
```
